```python
#Importacion de Librerias
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud

#Datos de referencia
comentarios=[
    "Me encanta este celular porque tiene buena bateria",
    "La pantalla del celular es muy mala",
    "Excelente servicio y excelente calidad",
    "No me gusto el producto porque llego dañado",
    "La bateria dura mucho tiempo y funciona muy bien",
    "Pesimo servicio al cliente",
    "Muy buena experiencia de compra",
    "No recomiendo este producto",
    "La calidad es excelente y el precio tambien",
    "El envio fue lento pero el producto es bueno"
]

df= pd.DataFrame({
    "Comentario": comentarios
})

print(df)

#Stopwords- palabras que no aportan mucho
stopword_es=[
    "el", "la", "los", "las",
    "de", "y", "a", "que",
    "porque", "muy", "pero",
    "al", "me", "no", "es"
]

#Bolsa de Palabras - Bag of words
vectorizador = CountVectorizer(
    stop_words= stopword_es
)

#Convertir texto a matriz numerica
X= vectorizador.fit_transform(df["Comentario"])

#Palabras detectadas
palabras = vectorizador.get_feature_names_out()
print(palabras)

#Matriz Baf of words
matrix_bow= pd.DataFrame(
  X.toarray(),
  columns= palabras
)
print("Matriz de bolsa de palabras", matrix_bow)

#Frecuencia Total de Palabras
frecuencias= matrix_bow.sum().sort_values(ascending=False)
print(frecuencias)

#Visualizacion- barras
frecuencias.head(10).plot(
    kind="bar",
    figsize=(10,5)
)
plt.title("Top 10 Palabras Mas frecuentes")
plt.xlabel("Palabras")
plt.ylabel("Frecuencia")
plt.xticks(rotation=45)
plt.show()

#Nube de Palabras
texto_total = " ".join(df["Comentario"])

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    stopwords=set(stopword_es)
).generate(texto_total)
plt.figure(figsize=(12,6))
plt.imshow(wordcloud)
plt.axis("off")
plt.title("Nube de palabras")
plt.show()


"""
comentarios = [
    "Me encanta este celular porque tiene buena bateria",
    "La pantalla del celular es muy mala",
    "Excelente servicio y excelente calidad",
    "No me gusto el producto porque llego dañado",
    "La bateria dura mucho tiempo y funciona muy bien",
    "Pesimo servicio al cliente",
    "Muy buena experiencia de compra",
    "No recomiendo este producto",
    "La calidad es excelente y el precio tambien",
    "El envio fue lento pero el producto es bueno",
    "Excelente atencion y excelente producto",
    "Muy mala experiencia con la compra",
    "La bateria del telefono es increible",
    "Pesimo envio y mala calidad",
    "Muy satisfecho con el servicio"
]


1- dataframe
2- funcion que realice limpieza:
- convertir a minusculas
- eliminar numero y caracteres especiales
- eliminar espacios extras
3- aplicar la funcion limpieza al dataframe
4- tokenizacion dataframe
"""



```

```python
# Importacion de Librerias
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud
import re # Import regular expression module for cleaning

# Datos de referencia - Ampliados
comentarios = [
    "Me encanta este celular porque tiene buena bateria",
    "La pantalla del celular es muy mala",
    "Excelente servicio y excelente calidad",
    "No me gusto el producto porque llego dañado",
    "La bateria dura mucho tiempo y funciona muy bien",
    "Pesimo servicio al cliente",
    "Muy buena experiencia de compra",
    "No recomiendo este producto",
    "La calidad es excelente y el precio tambien",
    "El envio fue lento pero el producto es bueno",
    "Excelente atencion y excelente producto",
    "Muy mala experiencia con la compra",
    "La bateria del telefono es increible",
    "Pesimo envio y mala calidad",
    "Muy satisfecho con el servicio"
]

df = pd.DataFrame({
    "Comentario": comentarios
})

print("DataFrame original:")
print(df)

# 2- Funcion para realizar limpieza de texto
def clean_text(text):
    # Convertir a minusculas
    text = text.lower()
    # Eliminar numeros y caracteres especiales (manteniendo solo letras y espacios)
    text = re.sub(r'[^a-zñáéíóúü\s]', '', text)
    # Eliminar espacios extras
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# 3- Aplicar la funcion de limpieza al DataFrame
df['Comentario_limpio'] = df['Comentario'].apply(clean_text)

print("\nDataFrame con comentarios limpios:")
print(df)

# Stopwords - palabras que no aportan mucho
stopword_es = [
    "el", "la", "los", "las",
    "de", "y", "a", "que",
    "porque", "muy", "pero",
    "al", "me", "no", "es",
    "un", "una", "unos", "unas", # Added more common stopwords
    "con", "sin", "para", "por",
    "en", "se", "mas", "como"
]

# 4- Tokenizacion y Bolsa de Palabras (Bag of Words) en comentarios limpios
vectorizador = CountVectorizer(
    stop_words=stopword_es
)

# Convertir texto limpio a matriz numerica
X = vectorizador.fit_transform(df["Comentario_limpio"])

# Palabras detectadas (vocabulario)
palabras = vectorizador.get_feature_names_out()
print("\nPalabras detectadas (vocabulario):")
print(palabras)

# Matriz Bag of Words
matrix_bow = pd.DataFrame(
    X.toarray(),
    columns=palabras
)
print("\nMatriz de bolsa de palabras:")
print(matrix_bow)

# Frecuencia Total de Palabras
frecuencias = matrix_bow.sum().sort_values(ascending=False)
print("\nFrecuencia total de palabras:\n")
print(frecuencias)

# Visualizacion - Barras
frecuencias.head(10).plot(
    kind="bar",
    figsize=(10, 5)
)
plt.title("Top 10 Palabras Mas Frecuentes (Despues de Limpieza)")
plt.xlabel("Palabras")
plt.ylabel("Frecuencia")
plt.xticks(rotation=45)
plt.show()

# Nube de Palabras en comentarios limpios
texto_total_limpio = " ".join(df["Comentario_limpio"])

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    stopwords=set(stopword_es)
).generate(texto_total_limpio)
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.title("Nube de Palabras (Despues de Limpieza)")
plt.show()
```
