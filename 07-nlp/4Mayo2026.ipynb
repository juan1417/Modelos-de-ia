```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from textblob import TextBlob
```

```python

#dataset de referencia
comentarios = [

    "Este servicio es ideal",
    "La calidad es superior",
    "Es un producto super",
    "Es una idea original",
    "Un resultado natural",
    "Terrible experiencia",
    "Un servicio horrible",
    "Un error fatal",
    "Un trato cruel",
    "Un acto criminal",
    "Es un objeto base",
    "El control es normal",
    "Es un articulo de gas",
    "El motor es gris",
    "La tienda esta en el centro"
]

df = pd.DataFrame({"comentario": comentarios})
```

```python
def analizar_sentimiento(texto):
  analisis = TextBlob(texto)
  polaridad = analisis.sentiment.polarity

  if polaridad > 0:
    return 'Positivo'
  elif polaridad == 0:
    return 'Neutro'
  else:
    return 'Negativo'
```

```python
#aplicar analisis
df['sentiminetos'] = df['comentario'].apply(analizar_sentimiento)
print(df)
```

```python
#conteo de centimientos

conteo = df['sentiminetos'].value_counts()
conteo.plot(kind="bar")
plt.title("Distribuvion de sentimiento")
plt.xlabel("Sentimineto")
plt.ylabel("Cantidad")
plt.show()
```

```python
#nube de palabras
text_total = " ".join(df['comentario'])
wordcloud = WordCloud(width=800, height=400).generate(text_total)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Nube de palabras")
plt.show()
```
